```python
import json

# === 1. 模拟输入：Agent 从网络上抓取的原始脏数据 ===
# 场景：模拟 Tavily 搜索 API 返回的结果，包含空格、字符串类型的数字等“脏”数据
raw_response = """
{
    "source": "Tavily_Search_API",
    "timestamp": "2026-01-26",
    "results": [
        {"model": "DeepSeek-V3", "cost": 10, "currency": "CNY"},
        {"model": "GPT-4o", "cost": "200", "currency": "USD"},
        {"model": "  Llama-3  ", "cost": 0, "currency": "OpenSource"}
    ]
}
"""

# === 2. 定义处理函数：Agent 的“消化系统” ===
def parse_agent_data(json_str):
    """
    功能：将脏 JSON 字符串清洗为干净的 Python 列表
    核心逻辑：解码 -> 防御性读取 -> 数据清洗 -> 类型转换
    """
    try:
        # A. 解码 (Decoder)：把死字符串变成活字典
        data = json.loads(json_str)
        print(f"✅ 数据来源: {data.get('source')}")

        clean_list = []

        # B. 循环处理 (Loop)：遍历每一条结果
        # 使用 .get() 防止 results 字段不存在时报错
        for item in data.get("results", []):

            # C. 清洗字符串 (Trimming)：去除前后空格
            # .get("model", "Unknown") 意思是：取不到 model 就用 "Unknown" 代替
            name = item.get("model", "Unknown").strip()

            # D. 类型转换 (Type Conversion)：把字符串数字变整数
            raw_cost = item.get("cost")
            if isinstance(raw_cost, str):
                cost = int(raw_cost) # 强转： "200" -> 200
            else:
                cost = raw_cost

            # E. 打包成品
            clean_list.append({"name": name, "price": cost})

        return clean_list

    # F. 熔断机制 (Fuse)：专门捕获 JSON 格式错误
    except json.JSONDecodeError:
        print("❌ 严重错误：抓取到的不是合法的 JSON！")
        return []

    # G. 兜底机制：捕获其他所有未知错误
    except Exception as e:
        print(f"⚠️ 发生未知错误: {e}")
        return []

# === 3. 运行测试 ===
print("--- 开始清洗数据 ---")
output = parse_agent_data(raw_response)
print("\n--- 清洗结果 ---")
print(output)
```
